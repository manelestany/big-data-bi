#Tarea 5 · Un dato semiestructurado: un pedido online
Este es un pedido real de la tienda online, escrito como diccionario de Python. Cópialo entero en una celda.
Responde con código, una línea por pregunta:

¿En qué provincia vive el cliente?
¿Cuántas líneas tiene el pedido?
¿Cómo se llama el segundo producto?
¿Qué tipo de entrega tiene?
¿Tiene la primera línea un campo descuento_pct? ¿Y la segunda? Pregúntalo con in.
Añade al pedido una clave "revisado" con el valor True.
Calcula cuánto paga el cliente en total, sumando el coste de envío.

```python
pedido = {
    "pedido_id": "PED0170755",
    "fecha": "2025-01-15T20:20:27",
    "cliente": {"id": "C01384", "provincia": "Madrid", "tiene_app": True},
    "tienda_id": "T006",
    "lineas": [
        {"producto_id": "P0165", "nombre": "Zumo de piña 1 L Colmena", "categoria": "Bebidas",
         "unidades": 1, "precio": 1.11, "descuento_pct": 5},
        {"producto_id": "P0101", "nombre": "Kéfir 500 ml Colmena", "categoria": "Lácteos y huevos",
         "unidades": 1, "precio": 1.38},
        {"producto_id": "P0268", "nombre": "Cacao soluble 500 g El Molinillo", "categoria": "Despensa",
         "unidades": 1, "precio": 3.53},
    ],
    "total": 5.96,
    "entrega": {"tipo": "domicilio", "cp": "28037", "franja": "tarde", "coste_envio": 4.95},
    "pago": {"metodo": "tarjeta"},
    "estado": "entregado",
    "valoracion": 3,
}

print(pedido["cliente"]["provincia"])
print(len(pedido["lineas"]))
print(pedido["lineas"][1]["nombre"])
print(pedido["entrega"]["tipo"])
print("descuento_pct" in pedido["lineas"][0])
print("descuento_pct" in pedido["lineas"][1])
pedido["revisado"]=True
total_pedido=pedido["total"]+pedido["entrega"]["coste_envio"]
print(f"total con envio: {total_pedido} e")
```

Este diccionario es un dato semiestructurado porque no esta ordenado en una tabla pero sigue una organización interna y es más flexible (el "descuento_pct" no lo tienen todas las lineas del pedido) y tiene diccionarios dentro de lineas.

#Tarea 6 · De texto JSON a diccionario
En un archivo, un pedido no es un diccionario: es una línea de texto. Esta es otra línea real del archivo de pedidos:
Muestra de qué tipo es texto.
Conviértelo en un diccionario llamado pedido2 con json.loads. Recuerda escribir antes import json.
Muestra de qué tipo es pedido2, su total y su tipo de entrega.
Pregunta con in si la entrega tiene código postal ("cp") y si el pedido tiene "valoracion".

```python
texto = '{"pedido_id": "PED0167736", "fecha": "2025-01-01T22:22:59", "cliente": {"id": "C00542", "provincia": "Albacete", "tiene_app": true}, "tienda_id": "T097", "lineas": [{"producto_id": "P0004", "nombre": "Tomate de ensalada 1 kg", "categoria": "Frutas y verduras", "unidades": 1, "precio": 2.61}, {"producto_id": "P0204", "nombre": "Arroz redondo 1 kg Colmena", "categoria": "Despensa", "unidades": 2, "precio": 1.11}], "total": 4.83, "entrega": {"tipo": "recogida en tienda"}, "pago": {"metodo": "tarjeta"}, "estado": "entregado"}'

print(type(texto))
import json
pedido2=json.loads(texto)
print(type(pedido2))
print(pedido2["total"])
print(pedido2["entrega"]["tipo"])
print("cp" in pedido2["entrega"])
print("valoracion" in pedido2)
```

#Tarea 7 · Un dato no estructurado: una reseña
¿Cuántos caracteres tiene? ¿Y cuántas palabras?
Pártela en frases cortando por ". ", un punto seguido de un espacio. ¿Cuántas salen?
¿Aparece la palabra cola?
Muestra la segunda frase.
Muestra la primera frase con la inicial en mayúscula. Busca cómo hacerlo: hay una herramienta de los textos que se llama capitalize.

```python
resena = "poca variedad en la carnicería. demasiada cola y solo dos cajas abiertas. no vuelvo."

print(len(resena))
total_palabras=len(resena.split())
print(total_palabras)
resena2=resena.split(". ")
print(len(resena.split(". ")))
print("cola" in resena)
print(resena2[1])
print(resena2[0].capitalize())
```
